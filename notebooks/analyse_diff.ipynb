```python
# calculate the ground truth distribution of yes, no, maybe answers from any one of the files
yes_count = 0
no_count = 0
maybe_count = 0
total_count = len(vanilla_data)
for i in range(total_count):
    answer = vanilla_data[i]['final_decision'].strip().lower()
    if 'yes' in answer:
        yes_count += 1
    elif 'no' in answer:
        no_count += 1
    else:
        maybe_count += 1


print("num of yeses, and %", yes_count, yes_count/total_count)
print("num of nos, and %", no_count, no_count/total_count)
print("num of maybes, and %", maybe_count, maybe_count/total_count)
```
