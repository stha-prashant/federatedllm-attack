# Rank of the different projections

```python
projections = {
    'base': 'project_matrix_safelora_torch.float32.pkl',
    'base_correct': 'project_matrix_safelora_torch.float32_correct.pkl',
    'harmful': 'project_matrix_safelora_torch.float32_harmful_systemprompt_backup.pkl',
    'harmful_correct': 'project_matrix_safelora_torch.float32_harmful_systemprompt_correct.pkl',
}
import numpy as np
import pickle
from tqdm import tqdm
ranks = {}
for projection_name, path in projections.items():
    ranks[projection_name] = []
    with open(path, 'rb') as f:
        ps = pickle.load(f)
    for p in tqdm(ps):
        rank = np.linalg.matrix_rank(p)
        ranks[projection_name].append(rank)
```

```python
# save to pkl
with open('./save_ranks_projections', 'wb') as file:
    pickle.dump(ranks, file)
```
