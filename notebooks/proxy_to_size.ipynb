```python
import os
import json
import matplotlib.pyplot as plt

path = "/scratch/ps9044/newsetting/PubMedQA7_BeaverTails3_1.00_1.00_1.00_1.00_0.50_0.50_0.50_200_safe_lora_c7s10_i10_b16a1_l512_r32a64_20260218140242/safelora"

malicious_idx = [0, 1, 2, 3]
benign_idx = [4, 5, 6]

rounds = []
all_scores = []

for r in range(30):
    file_path = os.path.join(path, f"round_{r}.json")
    with open(file_path, "r") as f:
        data = json.load(f)
    scores = data["S_total"]   # list of 7 scores
    rounds.append(r)
    all_scores.append(scores)

# plot
plt.figure(figsize=(10, 6))

for idx in malicious_idx:
    y = [scores[idx] for scores in all_scores]
    plt.scatter(rounds, y, color="red", label="malicious" if idx == malicious_idx[0] else None)
    for x, yy in zip(rounds, y):
        plt.text(x, yy, str(idx), fontsize=8, color="red")

for idx in benign_idx:
    y = [scores[idx] for scores in all_scores]
    plt.scatter(rounds, y, color="blue", label="benign" if idx == benign_idx[0] else None)
    for x, yy in zip(rounds, y):
        plt.text(x, yy, str(idx), fontsize=8, color="blue")

plt.xlabel("Round")
plt.ylabel("S_total score")
plt.title("Scatter plot of client scores across rounds")
plt.xticks(rounds)
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```
